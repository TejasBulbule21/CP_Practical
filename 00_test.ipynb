```python
# get mark of a student and give grade based on the mark
# 95 and above - A, 85 and above - B, 75 and above - C, 65 and above - D
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("Grade: A")
elif marks >= 85:
    print("Grade: B")
elif marks >= 75:
    print("Grade: C")
elif marks >= 65:
    print("Grade: D")
else:
    print("Grade: F")
```
