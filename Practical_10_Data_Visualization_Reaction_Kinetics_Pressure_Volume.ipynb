## 2. Why Data Visualization is Important

Chemical engineers regularly work with experimental and process data. A table may contain many numbers, but a graph can make a relationship or trend much easier to see.

For example:

### Reaction kinetics
We may want to observe how concentration changes as reaction time increases.

### Pressure vs. Volume
We may want to observe how pressure changes when the volume of a gas changes.

### Engineering rule

A good plot should help answer a specific engineering question. It should have meaningful labels, units and a clear title.

## 4. Activity 2 – Reaction Kinetics Data

Consider the following **illustrative reaction kinetics data**. It represents the concentration of a reactant measured at different times.

Create a Pandas DataFrame containing:

- `Time_min`
- `Concentration_mol_L`

These values are provided only for programming and visualization practice.

````python
kinetics_df = pd.DataFrame({
    'Time_min': [0, 5, 10, 15, 20, 25, 30, 40, 50],
    'Concentration_mol_L': [1.00, 0.88, 0.77, 0.68, 0.60, 0.53, 0.47, 0.37, 0.29]
})

kinetics_df
````

## 5. Activity 3 – Plot Reaction Kinetics

Create a **line plot** of concentration against reaction time.

Use:

- **X-axis:** Time (min)
- **Y-axis:** Concentration (mol/L)

Your graph must contain:

- suitable title,
- axis labels,
- units,
- grid,
- appropriate figure size.

### Think

What happens to the concentration as time increases?

````python
# Activity 3
# Create the reaction kinetics concentration-versus-time plot.
plt.plot(kinetics_df['Time_min'], kinetics_df['Concentration_mol_L'], marker='o')
plt.title('Reaction Kinetics: Concentration vs Time')
plt.xlabel('Time (min)')
plt.ylabel('Concentration (mol/L)') 
plt.savefig
````

## 6. Activity 4 – Compare Two Reaction Runs

Suppose the same reaction is performed under a second condition.

Create a second concentration dataset and plot both reaction runs on the **same graph**.

Use a legend such as:

- Run 1
- Run 2

### Questions

1. Which run shows a faster decrease in concentration?
2. At 20 minutes, which run has the lower concentration?
3. Why is it useful to show both runs on the same graph?

````python
# Activity 4
# Add a second reaction run and plot both runs together.
plt.figure(figsize=(8, 6))
plt.plot(kinetics_df['Time_min'], kinetics_df['Concentration_mol_L'] * 0.9, marker='s', linestyle='--', label='Run 2')
plt.title('Reaction Kinetics: Concentration vs Time (Two Runs)')
plt.xlabel('Time (min)')
plt.ylabel('Concentration (mol/L)')
plt.legend()
plt.grid()
plt.savefig

````

````python
plt.figure(figsize=(8, 6))
plt.plot(kinetics_df['Time_min'], kinetics_df['Concentration_mol_L'], marker='o', linestyle='-', label='Run 1')
plt.plot(kinetics_df['Time_min'], kinetics_df['Concentration_mol_L'] * 0.9, marker='s', linestyle='--', label='Run 2')
plt.title('Reaction Kinetics: Concentration vs Time (Two Runs)')
plt.xlabel('Time (min)')
plt.ylabel('Concentration (mol/L)')
plt.legend()
plt.grid()
plt.savefig
````

## 7. Activity 5 – Pressure vs. Volume Data

Now work with the second application explicitly mentioned in the syllabus: **pressure vs. volume**.

Consider this simplified teaching dataset for a gas system:

- `Volume_L`
- `Pressure_bar`

The values are illustrative and are intended only to demonstrate data visualization.

````python
pv_df = pd.DataFrame({
    'Volume_L': [1.0, 1.2, 1.5, 1.8, 2.0, 2.5, 3.0, 3.5, 4.0],
    'Pressure_bar': [4.00, 3.34, 2.68, 2.23, 2.02, 1.62, 1.35, 1.15, 1.01]
})

pv_df
````

## 8. Activity 6 – Plot Pressure vs. Volume

Create a graph with:

- **X-axis:** Volume (L)
- **Y-axis:** Pressure (bar)

Use a suitable plot for the data.

Include:

- title,
- axis labels,
- units,
- grid,
- appropriate figure size.

### Think

Does pressure increase or decrease as volume increases?

````python
# Activity 6
# Create the pressure-versus-volume plot.
plt.figure(figsize=(8, 6))
plt.plot(pv_df['Volume_L'], pv_df['Pressure_bar'] * 0.9, marker='s', linestyle='--', label='Run 2')
plt.title('Reaction Kinetics: Pressure vs Volume (Two Runs)')
plt.xlabel('Volume (L)')
plt.ylabel('Pressure (bar)')
plt.grid()
plt.savefig
````

## 9. Activity 7 – Line Plot vs Scatter Plot

Create the pressure-volume graph again using a **scatter plot**.

Compare the line plot and scatter plot.

### Think

Which representation makes the individual measurements more visible?

Which representation makes the overall relationship easier to follow?

````python
# Activity 7
# Create a scatter plot for the pressure-volume measurements.
plt.figure(figsize=(8, 6))
plt.scatter(pv_df['Volume_L'], pv_df['Pressure_bar'] * 0.9, marker='o', linestyle='--', label='Run 2')
plt.title('Reaction Kinetics: Pressure vs Volume (Two Runs)')
plt.xlabel('Volume (L)')
plt.ylabel('Pressure (bar)')
plt.grid()
plt.savefig
````

## 10. Activity 8 – Add a Reference Relationship

For an idealized constant-temperature gas relationship, pressure and volume are inversely related.

Create a simple reference curve using a constant `k`:

```text
P = k / V
```

Use the experimental/teaching pressure-volume points and plot the reference relationship on the same graph.

### Important

The purpose here is to learn how to compare measured points with a simple mathematical relationship. Do not claim that real experimental data must exactly follow the idealized relationship.

````python
# Activity 8
# Plot the pressure-volume data together with a simple P = k/V reference curve.
plt.figure(figsize=(8, 6))
plt.plot(pv_df['Volume_L'], pv_df['Pressure_bar'] * 0.9, marker='s', linestyle='--', label='Run 2')
#plot a reference curve P = k/V, where k is a constant
k = 6 # constant for the reference curve
plt.plot(pv_df['Volume_L'], k / pv_df['Volume_L'], color='red', linestyle='-', label='Reference Curve: P = k/V')
plt.title('Reaction Kinetics: Pressure vs Volume (Two Runs)')
plt.xlabel('Volume (L)')
plt.ylabel('Pressure (bar)')
plt.grid()
plt.savefig
````

## 12. Activity 10 – Visualize Both Applications

Create two final plots in separate cells:
